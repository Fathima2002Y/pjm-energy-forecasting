# ARIMA and SARIMA Models – PJM Hourly Energy Forecast (P679)

**My part of the project:** build the ARIMA model and the SARIMA model and generate their forecasts.
**Not in this notebook (other team members):** XGBoost, LSTM, model evaluation, ARIMA-vs-SARIMA comparison.

### What the existing EDA already told us (used here, not repeated)
| Item | Finding in the EDA notebook |
|---|---|
| Date/time column | `Datetime` |
| Target column | `PJMW_MW` (megawatts) |
| Original frequency | hourly, 143,206 raw rows (2002-04-01 to 2018-08-03) |
| Cleaning done | no NaNs; 4 duplicate timestamps removed (keep first); 30 missing hours (mostly daylight-saving) left un-imputed; first 168 rows dropped (lag/rolling features) -> 143,034 rows |
| Stationarity (EDA) | ADF on the *hourly* series: p = 0.0 -> stationary. Not tested on the daily series |
| Trend | no clear long-term trend (yearly averages just fluctuate) |
| Seasonality | daily cycle (min ~4 AM, peak ~7-8 PM), weekly cycle (weekdays > weekends), yearly cycle (winter/summer high) |
| ACF/PACF | done on the hourly series only |
| Series prepared for ARIMA/SARIMA | **daily average** of the hourly data, missing days interpolated -> 5,962 days; **last 365 days = test set**, 5,597 days = train |
| Order suggested in EDA | ARIMA(3,1,3), with no supporting evidence shown |

### Workflow in simple words
1. Load the data and rebuild the *same* daily series/train-test split that the EDA used (so my forecasts line up with my teammates' work).
2. Check stationarity and seasonality on the **daily training data** (the missing piece in the EDA).
3. Decide `d`, `s`, `D` from tests; tune only `p, q` (and `P, Q`) with a small grid using **training data only**.
4. Fit final ARIMA and SARIMA on the training data -> forecast the 365 test days (for the evaluator).
5. Refit on all data -> forecast the **next 30 days** (business objective).

> **Note:** all decisions are computed by the code and printed, so the numbers you see when you run it are the real choices. Run the cells top to bottom.

## Cell 1 – Import libraries
**What/Why:** load the tools we need. Only pandas, numpy, matplotlib and statsmodels – nothing exotic.
**Code meaning:** `ARIMA` and `SARIMAX` are the model classes (SARIMAX with no extra regressors *is* SARIMA). `adfuller` is the stationarity test. `acf/pacf` measure how today relates to past days.
**Expected output:** "Libraries imported."
**Presentation:** "I used pandas for data handling and statsmodels for ARIMA/SARIMA."

```python
import warnings
import time
import itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX

warnings.filterwarnings("ignore")   # hide long convergence warnings to keep output readable
print("Libraries imported.")
```

## Cell 2 – Load the data
**What/Why:** read the same Excel file the EDA used. Keep `PJMW_MW_Hourly.xlsx` in the same folder as this notebook (the EDA notebook also uses this relative path).
**Expected output:** shape `(143206, 2)`, columns `Datetime`, `PJMW_MW`, no missing values.
**Presentation:** "The raw data is hourly PJM-West consumption in MW from 2002 to 2018."

```python
df = pd.read_excel("PJMW_MW_Hourly.xlsx")

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print(df.dtypes)
print("\nMissing values:\n", df.isnull().sum())
df.head()
```

## Cell 3 – Rebuild the EDA-prepared time series
**What:** repeat, in a few lines, exactly what the EDA already did – no new analysis:
1. fix data types, sort by time,
2. drop the 4 duplicate timestamps (keep first),
3. drop the first 168 rows (the EDA lost these when creating the 168-hour lag/rolling features; we do the same so our data is identical to the team's),
4. average to **daily** values and interpolate any missing day (EDA section "Create daily time series"),
5. keep the **last 365 days as test**, everything before as train.

**Why:** ARIMA/SARIMA on 143k hourly points would be extremely slow, and the EDA already decided to use the daily average.
**Expected output:** 143,034 hourly rows, 5,962 days, train 5,597 days (2002-04-08 to 2017-08-03), test 365 days (2017-08-04 to 2018-08-03). The check lines should say `True`.
**Presentation:** "I used exactly the daily series and 365-day test split prepared in the EDA so all models are comparable."

```python
# 1. types + sorting
df["Datetime"] = pd.to_datetime(df["Datetime"], errors="coerce")
df["PJMW_MW"] = pd.to_numeric(df["PJMW_MW"], errors="coerce")
df = df.dropna(subset=["Datetime", "PJMW_MW"]).sort_values("Datetime")

# 2. duplicate timestamps (daylight-saving repeats) - same rule as EDA
df = df.drop_duplicates(subset=["Datetime"], keep="first").reset_index(drop=True)

# 3. first 168 rows were removed in the EDA by the lag/rolling features
df = df.iloc[168:].reset_index(drop=True)
print("Hourly rows after EDA cleaning:", len(df), "(EDA: 143034) ->", len(df) == 143034)

# 4. hourly -> daily average (same as EDA)
hourly = df.set_index("Datetime")["PJMW_MW"]
daily = hourly.resample("D").mean().interpolate(method="time")
hours_per_day = hourly.resample("D").count()      # how many hourly readings each day has

# 5. chronological split: last 365 days = test (no shuffling!)
test_size = 365
train_daily = daily.iloc[:-test_size]
test_daily = daily.iloc[-test_size:]

print("Daily observations:", len(daily), "(EDA: 5962) ->", len(daily) == 5962)
print("Train:", len(train_daily), train_daily.index.min().date(), "to", train_daily.index.max().date(), "->", len(train_daily) == 5597)
print("Test :", len(test_daily), test_daily.index.min().date(), "to", test_daily.index.max().date())
```

## Cell 4 – Identify frequency and target
**What/Why:** confirm what we are modelling and define the dates we must forecast.
**Also checked (extra, small):** the number of hourly readings behind the first and last day. In the EDA data the last day (2018-08-03) should have only the 00:00 reading, so its "daily average" is not a real full-day average. I keep the EDA series unchanged for the **test-period forecast** (so it matches the evaluator's test set), but for the **future 30-day forecast** I fit only up to the last *complete* day so a one-hour "day" does not distort the forecast. Please tell the team about this.
**Expected output:** target `PJMW_MW`, frequency `D`, a plot of the daily series, future dates 2018-08-04 to 2018-09-02.
**Presentation:** "The target is daily average MW; the series is regular daily, from 2002 to 2018."

```python
target_name = "PJMW_MW"
print("Target column     :", target_name, "(average MW per day)")
print("Original frequency: hourly")
print("Modelling freq    :", pd.infer_freq(daily.index), "(D = daily)")
print("Observations      :", len(daily))

# how many hourly readings are behind the first / last day?
print("\nHourly readings in first day:", hours_per_day.iloc[0], "| last day:", hours_per_day.iloc[-1])
last_complete_day = hours_per_day[hours_per_day >= 20].index.max()
print("Last complete day:", last_complete_day.date())

# the 30 days AFTER the last date in the data
future_dates = pd.date_range(daily.index.max() + pd.Timedelta(days=1), periods=30, freq="D")
print("Future forecast window:", future_dates[0].date(), "to", future_dates[-1].date())

plt.figure(figsize=(14, 4))
plt.plot(train_daily, label="Train")
plt.plot(test_daily, label="Test period (not used for fitting/tuning)")
plt.title("Daily average PJM energy consumption")
plt.ylabel("MW"); plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

## Cell 5 – Stationarity, differencing and seasonality (daily training data)
**What:** the EDA only ran ADF on the hourly series (stationary, p = 0.0) and ACF/PACF on hourly data. ARIMA here works on the **daily** series, so I re-run the ADF test and ACF/PACF once on the daily **training** data. This is the only extra analysis.
**Why:** ARIMA needs a stationary series (stable mean/variance). ADF null hypothesis = "not stationary"; p <= 0.05 means stationary -> `d = 0`, otherwise difference once -> `d = 1`.
**Rule used in code:** `d = 0` if ADF p <= 0.05 else `d = 1`. I do not copy the EDA's `(3,1,3)`; the test decides.
**Expected output:** ADF results, chosen `d`, ACF/PACF plots. In the ACF you should see a repeating bump every 7 lags (weekly pattern).
**Presentation:** "Stationarity was tested with ADF on the daily training data. The p-value decides the differencing order `d`."

```python
def run_adf(series, label):
    result = adfuller(series.dropna(), autolag="AIC")
    print(f"ADF on {label}: statistic = {result[0]:.3f}, p-value = {result[1]:.5f}")
    return result[1]

p_level = run_adf(train_daily, "daily train (original)")
p_diff = run_adf(train_daily.diff(), "daily train (1st difference)")

d = 0 if p_level <= 0.05 else 1
print("\n-> Chosen differencing order d =", d)
if d == 0:
    print("   Series is already stationary, no regular differencing needed.")

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
plot_acf(train_daily, lags=40, ax=axes[0], title="ACF - daily train")
plot_pacf(train_daily, lags=40, ax=axes[1], method="ywm", title="PACF - daily train")
plt.tight_layout(); plt.show()
```

## Cell 6 – Explain and determine ARIMA (p, d, q)
**ARIMA in simple words:** it predicts tomorrow from (a) the past few days' values and (b) the past few forecast mistakes, after making the data stable.

`ARIMA(p, d, q)`
* **p (AR)** – how many previous days' values are used. "Today depends on the last p days."
* **d (I)** – how many times we subtract yesterday from today to make the series stable. `d = 0` means no differencing.
* **q (MA)** – how many previous forecast errors are used. "Correct today's forecast using the last q mistakes."

**How I decide the parameters (not blind):**
* `d` – from the ADF test in Cell 5.
* `p`, `q` – the PACF suggests `p`, the ACF suggests `q`. On daily data many lags look significant (the weekly pattern leaks in), so the plots alone cannot give a single answer -> **tuning is needed** (next cell).
* The EDA's `(3,1,3)` was stated without evidence, so I treat it only as a hint for the search range (0 to 3).

**Expected output:** the significant lags and the search grid.
**Presentation:** "ACF/PACF narrow down the range, and a small grid search with AIC picks the final p and q."

```python
n = len(train_daily)
band = 1.96 / np.sqrt(n)          # 95% significance band for ACF/PACF
base = train_daily.diff().dropna() if d == 1 else train_daily

pacf_vals = pacf(base, nlags=10, method="ywm")
acf_vals = acf(base, nlags=10)
print("Significant PACF lags (suggest p):", [k for k in range(1, 11) if abs(pacf_vals[k]) > band])
print("Significant ACF lags  (suggest q):", [k for k in range(1, 11) if abs(acf_vals[k]) > band])

# keep the search small and cheap: p and q from 0 to 3, d is fixed by the ADF test
p_values = range(0, 4)
q_values = range(0, 4)
print("\nARIMA search grid: p in", list(p_values), ", d =", d, ", q in", list(q_values),
      "->", len(p_values) * len(q_values), "models")
```

## Cell 7 – ARIMA hyperparameter tuning (needed: YES)
**Why tuning is needed:** ACF/PACF at daily level do not give one clear `(p, q)`, and the EDA gave no evidence for its `(3,1,3)`.
**How (small and manageable):** fit 16 models (`p, q` = 0..3, `d` fixed) on the **training data only** and pick the lowest **AIC** (AIC rewards fit and punishes complexity; lower is better). The test data is never used, and I don't vary `d` in the grid because AIC cannot be compared across different `d`.
**Expected output:** a table of the 5 best models and the chosen order. Takes a few minutes.
**Presentation:** "I tuned ARIMA with a 4x4 grid on training data only, using AIC to avoid overfitting."

```python
arima_results = []
start = time.time()

for p, q in itertools.product(p_values, q_values):
    try:
        model = ARIMA(train_daily, order=(p, d, q))
        fitted = model.fit()
        arima_results.append({"order": (p, d, q), "AIC": fitted.aic})
    except Exception as e:
        print("Skipped", (p, d, q), "->", e)

arima_table = pd.DataFrame(arima_results).sort_values("AIC").reset_index(drop=True)
best_arima_order = arima_table.loc[0, "order"]

print(f"Tuning finished in {(time.time() - start) / 60:.1f} minutes")
display(arima_table.head(5))
print("Selected ARIMA order (p,d,q):", best_arima_order)
```

## Cell 8 – Build the final ARIMA model
**What:** fit ARIMA with the selected order on the full **training** period (2002-04-08 to 2017-08-03).
**Expected output:** the statsmodels summary (coefficients, AIC).
**Presentation:** "The final ARIMA model is trained only on data before the test year."

```python
arima_model = ARIMA(train_daily, order=best_arima_order)
arima_fit = arima_model.fit()
print(arima_fit.summary())
```

## Cell 9 – ARIMA forecasts
**What:** two forecasts from the same ARIMA order:
1. **Test-period forecast** – model trained on train data only, predicts the 365 test dates (2017-08-04 to 2018-08-03). This is handed to the evaluator. I do **not** compare it with actual values here.
2. **Next-30-days forecast** – model refitted on all data up to the last complete day, predicts 2018-08-04 to 2018-09-02.

Each table has the forecast plus a 95% confidence range. Both are saved as CSV files next to the notebook.
**Expected output:** two plots and two CSV files: `arima_forecast_test_period.csv`, `arima_forecast_next_30_days.csv`.
**Note:** a non-seasonal ARIMA cannot repeat the weekly cycle, so its forecast will look smooth/flat for long horizons. That is normal ARIMA behaviour.
**Presentation:** "ARIMA gives a forecast and a confidence range for the test year and for the next 30 days."

```python
def make_forecast_table(fitted_result, steps):
    """Return forecast + 95% interval as a table."""
    fc = fitted_result.get_forecast(steps=steps)
    table = fc.predicted_mean.to_frame("forecast_MW")
    ci = fc.conf_int(alpha=0.05)
    table["lower_95"] = ci.iloc[:, 0]
    table["upper_95"] = ci.iloc[:, 1]
    return table

# ---- 1) test-period forecast (trained on train only)
arima_test_fc = make_forecast_table(arima_fit, steps=len(test_daily))
arima_test_fc.index = test_daily.index          # attach the test dates
arima_test_fc.index.name = "Date"
arima_test_fc.to_csv("arima_forecast_test_period.csv")

# ---- 2) next 30 days (refit on all data up to last complete day)
full_series = daily.loc[:last_complete_day]
arima_full_fit = ARIMA(full_series, order=best_arima_order).fit()
steps_needed = (future_dates[-1] - last_complete_day).days      # includes the skipped incomplete day
arima_future_fc = make_forecast_table(arima_full_fit, steps=steps_needed).loc[future_dates]
arima_future_fc.index.name = "Date"
arima_future_fc.to_csv("arima_forecast_next_30_days.csv")

print("ARIMA test-period forecast (first 5 rows):"); display(arima_test_fc.head())
print("ARIMA next-30-days forecast (first 5 rows):"); display(arima_future_fc.head())

fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].plot(train_daily.iloc[-180:], label="Train (last 180 days)")
axes[0].plot(arima_test_fc["forecast_MW"], label="ARIMA forecast")
axes[0].fill_between(arima_test_fc.index, arima_test_fc["lower_95"], arima_test_fc["upper_95"], alpha=0.2)
axes[0].set_title("ARIMA - test period forecast (365 days)"); axes[0].legend()
axes[1].plot(full_series.iloc[-90:], label="History (last 90 days)")
axes[1].plot(arima_future_fc["forecast_MW"], label="ARIMA next 30 days")
axes[1].fill_between(arima_future_fc.index, arima_future_fc["lower_95"], arima_future_fc["upper_95"], alpha=0.2)
axes[1].set_title("ARIMA - next 30 days"); axes[1].legend()
for a in axes:
    a.grid(alpha=0.3); a.set_ylabel("MW")
plt.tight_layout(); plt.show()
```

## Cell 10 – Explain SARIMA (p, d, q)(P, D, Q, s)
**SARIMA in simple words:** ARIMA + a second, "seasonal" ARIMA that looks back one whole season (e.g., the same weekday last week), so it can repeat a regular cycle.

| Symbol | Meaning |
|---|---|
| **p, d, q** | same as ARIMA: recent days, regular differencing, recent errors |
| **P** | how many *previous seasons* (e.g., same day last week, two weeks ago) the values depend on |
| **D** | seasonal differencing: subtract the value one season ago (`today - same day last week`) |
| **Q** | how many previous *seasonal* forecast errors are used |
| **s** | length of one season in observations (number of days in the repeating cycle) |

**Expected output:** the table below.
**Presentation:** "SARIMA is ARIMA plus a seasonal part; `s` is how many days one cycle lasts."

```python
sarima_explain = pd.DataFrame({
    "Parameter": ["p", "d", "q", "P", "D", "Q", "s"],
    "Meaning": ["recent values used", "regular differencing", "recent errors used",
                "past seasons used", "seasonal differencing", "past seasonal errors used",
                "season length (days)"],
    "How I decide": ["tuned (grid)", "ADF test (Cell 5)", "tuned (grid)",
                     "tuned (grid)", "seasonality-strength test (Cell 11)", "tuned (grid)",
                     "ACF peak + EDA weekly finding (Cell 11)"]
})
display(sarima_explain)
```

## Cell 11 – Determine the seasonal period `s` (and `D`) from the EDA/data
**Evidence from the EDA:** weekday vs weekend averages differ (5,755 vs 5,222 MW), the day-of-week chart shows a weekly cycle, and the hourly lag correlation at 168 hours (= 24 x 7 = one week) is high (0.76). Because our modelling series is **daily**, one week = **7 observations**. I do not simply assume it: the code measures which lag (2 to 30) has the strongest autocorrelation in the day-to-day changes.
**Yearly cycle (about 365 days):** the EDA shows a strong winter/summer effect, but SARIMA with `s = 365` is computationally unrealistic in statsmodels (huge memory/time), so I use the weekly cycle and mention the yearly effect as a limitation.
**Choosing `D`:** I measure how strong the seasonality is (share of variance explained by the repeating pattern). If strength > 0.64 (a widely used rule of thumb) -> `D = 1`, else `D = 0`.
**Expected output:** top lags (7, 14, 21 should stand out), chosen `s`, the strength value and `D`.
**Presentation:** "The EDA showed a weekday/weekend pattern and the ACF confirms a peak at lag 7, so s = 7 for daily data."

```python
# Look at day-to-day changes so the slow level does not hide the repeating pattern
changes = train_daily.diff().dropna()
acf_changes = acf(changes, nlags=60)

candidate_lags = range(2, 31)
top_lags = sorted(candidate_lags, key=lambda k: acf_changes[k], reverse=True)[:5]
print("Lags (2-30) with strongest autocorrelation:", top_lags)
s = top_lags[0]
print("-> Chosen seasonal period s =", s, "days")

plt.figure(figsize=(12, 3.5))
plt.bar(range(1, 61), acf_changes[1:], width=0.6)
plt.axhline(band, color="red", ls="--"); plt.axhline(-band, color="red", ls="--")
for k in range(s, 61, s):
    plt.axvline(k, color="green", alpha=0.3)
plt.title(f"ACF of daily changes (green lines every {s} lags)")
plt.xlabel("Lag (days)"); plt.tight_layout(); plt.show()

# seasonal differencing decision D
decomp = seasonal_decompose(train_daily, model="additive", period=s)
resid = decomp.resid.dropna()
seas = decomp.seasonal.loc[resid.index]
strength = max(0, 1 - resid.var() / (resid + seas).var())
D = 1 if strength > 0.64 else 0
print(f"Seasonal strength = {strength:.2f}  ->  D = {D}")
print(f"\nFixed so far: d = {d}, D = {D}, s = {s}")
```

## Cell 12 – SARIMA hyperparameter tuning (needed: YES)
**Why:** there are 4 free parameters (`p, q, P, Q`) and the plots cannot fix them.
**How (small):** `d`, `D`, `s` are already fixed. Search only `p, q` in {0,1,2} and `P, Q` in {0,1} = **36 models**, chosen by lowest **AIC**, on training data only.
**Speed trick:** to keep runtime manageable, the search uses only the **last 3 years of the training data** (1,095 days); the final model in Cell 13 is fitted on the full training data with the chosen orders.
**Expected output:** top-5 table and the selected order. Runtime: several minutes – be patient.
**Presentation:** "SARIMA tuning used a 36-model grid on training data with AIC; test data was never touched."

```python
tune_data = train_daily.iloc[-1095:]        # last 3 years of TRAINING data only

# a constant is needed only if there is no differencing at all
trend_term = "c" if (d == 0 and D == 0) else None

sarima_results = []
start = time.time()
for p, q, P, Q in itertools.product([0, 1, 2], [0, 1, 2], [0, 1], [0, 1]):
    try:
        model = SARIMAX(tune_data,
                        order=(p, d, q),
                        seasonal_order=(P, D, Q, s),
                        trend=trend_term,
                        enforce_stationarity=False,
                        enforce_invertibility=False)
        fitted = model.fit(disp=False, maxiter=50)
        sarima_results.append({"order": (p, d, q), "seasonal_order": (P, D, Q, s), "AIC": fitted.aic})
    except Exception as e:
        print("Skipped", (p, d, q), (P, D, Q, s), "->", e)

sarima_table = pd.DataFrame(sarima_results).sort_values("AIC").reset_index(drop=True)
best_sarima_order = sarima_table.loc[0, "order"]
best_sarima_seasonal = sarima_table.loc[0, "seasonal_order"]

print(f"Tuning finished in {(time.time() - start) / 60:.1f} minutes")
display(sarima_table.head(5))
print("Selected SARIMA (p,d,q)(P,D,Q,s):", best_sarima_order, best_sarima_seasonal)
```

## Cell 13 – Build the final SARIMA model
**What:** fit SARIMA with the selected orders on the **full training data** (all days before the test year).
**Expected output:** model summary (coefficients, AIC).
**Presentation:** "The final SARIMA has the weekly seasonal part added to ARIMA and is trained only on pre-test data."

```python
sarima_model = SARIMAX(train_daily,
                       order=best_sarima_order,
                       seasonal_order=best_sarima_seasonal,
                       trend=trend_term,
                       enforce_stationarity=False,
                       enforce_invertibility=False)
sarima_fit = sarima_model.fit(disp=False, maxiter=200)
print(sarima_fit.summary())
```

## Cell 14 – SARIMA forecasts
**What:** same two forecasts as ARIMA:
1. test-period forecast (365 days, trained on train only) -> for the evaluator,
2. next-30-days forecast (refit on all data up to the last complete day).

Saved as `sarima_forecast_test_period.csv` and `sarima_forecast_next_30_days.csv`. **No evaluation or comparison is done here.**
**Expected output:** two plots; the 30-day plot should show the weekly up-and-down pattern (lower weekends).
**Presentation:** "SARIMA forecasts keep the weekly rhythm, which plain ARIMA cannot. Evaluation is done by my teammate."

```python
# ---- 1) test-period forecast (trained on train only)
sarima_test_fc = make_forecast_table(sarima_fit, steps=len(test_daily))
sarima_test_fc.index = test_daily.index
sarima_test_fc.index.name = "Date"
sarima_test_fc.to_csv("sarima_forecast_test_period.csv")

# ---- 2) next 30 days (refit on all data up to last complete day)
sarima_full_fit = SARIMAX(full_series,
                          order=best_sarima_order,
                          seasonal_order=best_sarima_seasonal,
                          trend=trend_term,
                          enforce_stationarity=False,
                          enforce_invertibility=False).fit(disp=False, maxiter=200)
sarima_future_fc = make_forecast_table(sarima_full_fit, steps=steps_needed).loc[future_dates]
sarima_future_fc.index.name = "Date"
sarima_future_fc.to_csv("sarima_forecast_next_30_days.csv")

print("SARIMA test-period forecast (first 5 rows):"); display(sarima_test_fc.head())
print("SARIMA next-30-days forecast (first 5 rows):"); display(sarima_future_fc.head())

fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].plot(train_daily.iloc[-180:], label="Train (last 180 days)")
axes[0].plot(sarima_test_fc["forecast_MW"], label="SARIMA forecast")
axes[0].fill_between(sarima_test_fc.index, sarima_test_fc["lower_95"], sarima_test_fc["upper_95"], alpha=0.2)
axes[0].set_title("SARIMA - test period forecast (365 days)"); axes[0].legend()
axes[1].plot(full_series.iloc[-90:], label="History (last 90 days)")
axes[1].plot(sarima_future_fc["forecast_MW"], label="SARIMA next 30 days")
axes[1].fill_between(sarima_future_fc.index, sarima_future_fc["lower_95"], sarima_future_fc["upper_95"], alpha=0.2)
axes[1].set_title("SARIMA - next 30 days"); axes[1].legend()
for a in axes:
    a.grid(alpha=0.3); a.set_ylabel("MW")
plt.tight_layout(); plt.show()

print("Done. Hand the *_test_period.csv files to the evaluation team member.")
```

## Why these parameters (fill in the numbers your run printed)
* **d** – from ADF on the daily training data (Cell 5). If p <= 0.05 the series is stationary -> `d = 0`. (The EDA also found the hourly series stationary.)
* **ARIMA p, q** – lowest AIC in the 4x4 grid (Cell 7); ACF/PACF only gave the range.
* **s** – ACF of daily changes peaks at lag 7; matches the EDA's weekday/weekend finding and the 168-hour (24x7) lag. The yearly (365) cycle is not modelled because it is too heavy for SARIMA.
* **D** – seasonal-strength test (Cell 11).
* **SARIMA p, q, P, Q** – lowest AIC in the 36-model grid (Cell 12).

## Is tuning needed?
Yes for both. The EDA gave stationarity and seasonality hints but not `p, q, P, Q`, and daily ACF/PACF are cluttered by the weekly pattern. The grids are small (16 and 36 models), use training data only and use AIC, so they are cheap and leak-free. `d`, `D`, `s` come from tests, not from the search.

## Simple presentation script (about 2 minutes)
"The EDA converted hourly consumption to a daily average and kept the last 365 days as the test set. My job was ARIMA and SARIMA on that series. I first checked stationarity on the daily training data with the ADF test, which fixed the differencing order. ARIMA uses the previous days' values (p), differencing (d) and previous errors (q). Since ACF/PACF did not give one clear answer, I tuned p and q with a small grid using AIC on training data only. For SARIMA I added the seasonal part. The EDA showed weekday-versus-weekend patterns and the ACF peaks every 7 days, so the seasonal period is 7. I tuned p, q, P, Q on a 36-model grid. Both final models were trained before the test year and produced a 365-day forecast for evaluation and a 30-day forecast beyond August 2018. Evaluation and comparison are done by a teammate."

## Likely viva questions
1. **What is ARIMA?** A model that predicts using past values (AR), differencing (I) and past errors (MA).
2. **What is stationarity and why does it matter?** Stable mean/variance over time; ARIMA assumes it.
3. **How did you choose d?** ADF test on the daily training series.
4. **Why daily, not hourly data?** Hourly has 143k points with several cycles (24, 168, ~8760); ARIMA/SARIMA would be very slow. The team's EDA chose daily averages.
5. **How did you choose s?** ACF peak at lag 7 and the EDA weekday/weekend pattern. Not assumed.
6. **Why not s = 365?** Too heavy for SARIMA; the yearly effect is a stated limitation.
7. **What is AIC?** A score for fit vs complexity; lower is better.
8. **Why not use the test data in tuning?** It would leak the future and give unrealistic results.
9. **Why refit before the 30-day forecast?** To use the most recent data; the orders stay the same.
10. **Why does the ARIMA forecast look flat?** It has no seasonal part, so long forecasts drift to the average; SARIMA repeats the weekly cycle.
11. **Did you compare ARIMA and SARIMA?** No, evaluation is another team member's task.
12. **What is the data caveat?** The last day has only one hourly reading, so the 30-day forecast is fitted up to the last complete day.
