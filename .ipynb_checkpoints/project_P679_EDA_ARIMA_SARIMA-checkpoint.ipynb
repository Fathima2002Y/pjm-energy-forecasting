### FEATURE ENGINEERING

```python
# Creating lag features using previous values

df["Lag_1_Hour"] = df["PJMW_MW"].shift(1)
df["Lag_24_Hours"] = df["PJMW_MW"].shift(24)
df["Lag_48_Hours"] = df["PJMW_MW"].shift(48)
df["Lag_168_Hours"] = df["PJMW_MW"].shift(168)

# Creating rolling features using only previous values

df["Rolling_24H"] = df["PJMW_MW"].shift(1).rolling(24).mean()
df["Rolling_7D"] = df["PJMW_MW"].shift(1).rolling(168).mean()

print("Lag and rolling features created successfully.")
```

```python
# Removing rows with missing lag and rolling values

lag_rolling_cols = [
    "Lag_1_Hour",
    "Lag_24_Hours",
    "Lag_48_Hours",
    "Lag_168_Hours",
    "Rolling_24H",
    "Rolling_7D"
]

df = df.dropna(subset=lag_rolling_cols).reset_index(drop=True)

print("Dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
```

```python
# Checking missing values after lag and rolling feature creation

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

display(missing_summary)
```

```python
# Removing rows with missing lag/rolling features

required_features = [
    "Lag_1_Hour",
    "Lag_24_Hours",
    "Lag_48_Hours",
    "Lag_168_Hours",
    "Rolling_24H",
    "Rolling_7D"
]

df = df.dropna(subset=required_features).reset_index(drop=True)

print("Shape after removing lag/rolling NaN rows:", df.shape)
```

```python
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
#Extract the year from the datetime column.
df["Year"] = df["Datetime"].dt.year

#Extract the month number(1-12).
df["Month"] = df["Datetime"].dt.month

#Extract the month name.
df["Month_Name"] = df["Datetime"].dt.month_name()

#Extract the day of the month.
df["Day"] = df["Datetime"].dt.day

#extract the hour of the day.
df["Hour"] = df["Datetime"].dt.hour

#Extract the day of the week(Monday=0,sunday = 6)
df["Day_of_Week"] = df["Datetime"].dt.dayofweek

#Extract the day Name.
df["Day_Name"] = df["Datetime"].dt.day_name()

#Extract the quater of the year.
df["Quarter"] = df["Datetime"].dt.quarter

#Extract the iSO week number.
df["Week"] = df["Datetime"].dt.isocalendar().week.astype(int)

# Identifies whether the day is a weekend
#Saturday = 5 and sunday =6
df["Weekend"] = df["Day_of_Week"] >= 5
```

```python
# Define a function to assign a seasin based on the month.
def get_season(month):

#December,January,february are classified as winter.
    if month in [12, 1, 2]:
        return "Winter"
#March,April,May are classified as spring.
    elif month in [3, 4, 5]:
        return "Spring"
#June,July,August are classified as Summer.
    elif month in [6, 7, 8]:
        return "Summer"
#September,October,November are classified as Autum.
    else:
        return "Autumn"


df["Season"] = df["Month"].apply(get_season)

display(df.head())
```

### Outlier Analysis and Treatment

```python
#OUTLIER DETEECTION USING THE (IQR).
#Calculates the first and third quartile(25th percentile, 75th percentile)
Q1 = df["PJMW_MW"].quantile(0.25)
Q3 = df["PJMW_MW"].quantile(0.75)

#Calculates the interquartile Range.
IQR = Q3 - Q1

#Calculates the lower and upper limits for detecting outliers.
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

#Identify observations below the lower bound  or above the upper bound as potential outliers.
outliers = df[
    (df["PJMW_MW"] < lower_bound) |
    (df["PJMW_MW"] > upper_bound)
]

#Displays the calculated outliers boundries.
print("\nIQR Lower Bound:", lower_bound)
print("IQR Upper Bound:", upper_bound)

#Displays the number of potential outliers.
print("Potential Outliers:", len(outliers))
```

###Outlier Treatment:
Potential outliers were identified using the IQR method. Since electricity consumption can naturally exhibit high-demand and low-demand periods, these observations were retained because extreme values may represent genuine peak demand rather than data errors.

#### INFERENCE:
The IQR method was used to identify potential outliers in PJM energy consumption. The observations outside the calculated lower and upper bounds are flagged as potential outliers. These values may represent periods of unusually high or low electricity demand, so they should be investigated rather than removed automatically.

### ADF TEST

```python
## ADF Stationarity Test
```

```python
from statsmodels.tsa.stattools import adfuller

# Augmented Dickey-Fuller test
adf_result = adfuller(df["PJMW_MW"].dropna())

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Number of Lags Used:", adf_result[2])
print("Number of Observations Used:", adf_result[3])

if adf_result[1] <= 0.05:
    print("\nResult: The series is stationary.")
else:
    print("\nResult: The series is non-stationary.")
```

```python
import time
import itertools
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.statespace.sarimax import SARIMAX

warnings.filterwarnings("ignore")
print("Libraries ready.")
```

## Cell 2 – Load the EDA-prepared data
**What/Why:** no need to read the Excel again. The EDA notebook already holds the cleaned hourly `df` and the daily series `daily_avg`. I only check that they exist and match the EDA numbers.
**Expected output:** hourly rows 143034, daily days 5962, and `True` on the checks.
**Presentation:** "I continued directly from the EDA-cleaned data instead of re-cleaning."

```python
print("Hourly rows in EDA df :", len(df), "-> matches EDA (143034):", len(df) == 143034)
print("Daily series daily_avg:", len(daily_avg), "days -> matches EDA (5962):", len(daily_avg) == 5962)
print("Missing daily values  :", daily_avg.isnull().sum())
daily_avg.head()
```

## Cell 3 – Prepare the time-series index
**What/Why:** `daily_avg` already has a date index (from `resample("D")`). I create a working copy called `daily`, rebuild the train/test split from it (the same last-365-days rule as the EDA, no shuffling) and count how many hourly readings sit behind each day.
**Why the count:** to spot incomplete days. In the EDA data the last day (2018-08-03) should contain only the 00:00 reading, so its "daily average" is not a real full-day average.
**Expected output:** train 5,597 days (2002-04-08 to 2017-08-03), test 365 days (2017-08-04 to 2018-08-03).
**Presentation:** "The split is chronological: the last year is the test set and is never used for fitting or tuning."

```python
daily = daily_avg.copy()
daily.index = pd.DatetimeIndex(daily.index)

# same split rule as the EDA: last 365 days = test
train_daily = daily.iloc[:-365]
test_daily = daily.iloc[-365:]
print("Train:", len(train_daily), train_daily.index.min().date(), "to", train_daily.index.max().date(), "-> matches EDA (5597):", len(train_daily) == 5597)
print("Test :", len(test_daily), test_daily.index.min().date(), "to", test_daily.index.max().date())

# how many hourly readings are behind each day?
hours_per_day = df.set_index("Datetime")["PJMW_MW"].resample("D").count()
print("\nHourly readings on first day:", hours_per_day.iloc[0], "| on last day:", hours_per_day.iloc[-1])
```

## Cell 4 – Identify frequency and target
**What/Why:** confirm what we are modelling and fix the dates to forecast.
**Handling the incomplete last day:** the test-period forecast keeps the EDA series unchanged (so it matches what the evaluator uses). For the **future 30-day forecast** I fit only up to the last *complete* day, so a one-hour "day" does not distort it. Please mention this to the team.
**Expected output:** target `PJMW_MW`, frequency `D`, future window 2018-08-04 to 2018-09-02.
**Presentation:** "The target is daily average MW; the forecast window follows the last date in the data."

```python
print("Target column     : PJMW_MW (average MW per day)")
print("Original frequency: hourly")
print("Modelling freq    :", pd.infer_freq(daily.index), "(D = daily)")
print("Observations      :", len(daily))

last_complete_day = hours_per_day[hours_per_day >= 20].index.max()
print("Last complete day :", last_complete_day.date())

future_dates = pd.date_range(daily.index.max() + pd.Timedelta(days=1), periods=30, freq="D")
print("Future forecast window:", future_dates[0].date(), "to", future_dates[-1].date())
```

## Cell 5 – Stationarity, differencing and seasonality (daily training data)
**What:** the EDA ran ADF/ACF/PACF only on the hourly series. ARIMA here uses the **daily** series, so I run the ADF test and ACF/PACF once on the daily **training** data. This is the only extra analysis.
**Why:** ARIMA needs a stable (stationary) series. ADF: p <= 0.05 -> stationary -> `d = 0`; otherwise difference once -> `d = 1`.
**Note:** I use the test result, not the EDA's `(3,1,3)`.
**Expected output:** ADF results, chosen `d`, ACF/PACF plots (look for bumps every 7 lags = weekly pattern).
**Presentation:** "The ADF p-value on the daily training data decides the differencing order d."

```python
def run_adf(series, label):
    result = adfuller(series.dropna(), autolag="AIC")
    print(f"ADF on {label}: statistic = {result[0]:.3f}, p-value = {result[1]:.5f}")
    return result[1]

p_level = run_adf(train_daily, "daily train (original)")
p_diff = run_adf(train_daily.diff(), "daily train (1st difference)")

d = 0 if p_level <= 0.05 else 1
print("\n-> Chosen differencing order d =", d)
if d == 0:
    print("   Series is already stationary, no regular differencing needed.")

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
plot_acf(train_daily, lags=40, ax=axes[0], title="ACF - daily train")
plot_pacf(train_daily, lags=40, ax=axes[1], method="ywm", title="PACF - daily train")
plt.tight_layout(); plt.show()
```

## Cell 6 – Explain and determine ARIMA (p, d, q)
**ARIMA in simple words:** it predicts tomorrow from (a) the past few days' values and (b) the past few forecast mistakes, after making the data stable.

`ARIMA(p, d, q)`
* **p (AR)** – how many previous days' values are used.
* **d (I)** – how many times we subtract yesterday from today to make the series stable (`0` = none).
* **q (MA)** – how many previous forecast errors are used.

**How I decide:** `d` from the ADF test (Cell 5). `p` from PACF and `q` from ACF, but on daily data many lags look significant (the weekly pattern leaks in), so the plots give only a range -> **tuning is needed**. The EDA's `(3,1,3)` had no evidence, so it is just a hint for the range 0-3.
**Expected output:** significant lags and the search grid.
**Presentation:** "ACF/PACF give the range; a small AIC grid search picks the final p and q."

```python
n = len(train_daily)
band = 1.96 / np.sqrt(n)          # 95% significance band
base = train_daily.diff().dropna() if d == 1 else train_daily

pacf_vals = pacf(base, nlags=10, method="ywm")
acf_vals = acf(base, nlags=10)
print("Significant PACF lags (suggest p):", [k for k in range(1, 11) if abs(pacf_vals[k]) > band])
print("Significant ACF lags  (suggest q):", [k for k in range(1, 11) if abs(acf_vals[k]) > band])

p_values = range(0, 4)
q_values = range(0, 4)
print("\nARIMA search grid: p in", list(p_values), ", d =", d, ", q in", list(q_values),
      "->", len(p_values) * len(q_values), "models")
```

## Cell 7 – ARIMA hyperparameter tuning (needed: YES)
**Why needed:** the plots do not give one clear `(p, q)` and the EDA gave no evidence for its `(3,1,3)`.
**How (small):** fit 16 models (`p, q` = 0..3, `d` fixed) on the **training data only**; choose the lowest **AIC** (rewards fit, penalises complexity; lower is better). `d` is not varied because AIC cannot be compared across different `d`. Test data is never used.
**Expected output:** top-5 table and the chosen order (takes a few minutes).
**Presentation:** "I tuned ARIMA with a 4x4 grid on training data using AIC."

```python
arima_results = []
start = time.time()

for p, q in itertools.product(p_values, q_values):
    try:
        fitted = ARIMA(train_daily, order=(p, d, q)).fit()
        arima_results.append({"order": (p, d, q), "AIC": fitted.aic})
    except Exception as e:
        print("Skipped", (p, d, q), "->", e)

arima_table = pd.DataFrame(arima_results).sort_values("AIC").reset_index(drop=True)
best_arima_order = arima_table.loc[0, "order"]

print(f"Tuning finished in {(time.time() - start) / 60:.1f} minutes")
display(arima_table.head(5))
print("Selected ARIMA order (p,d,q):", best_arima_order)
```

## Cell 8 – Build the final ARIMA model
**What:** fit ARIMA with the selected order on the whole **training** period.
**Expected output:** model summary.
**Presentation:** "The final ARIMA is trained only on data before the test year."

```python
arima_fit = ARIMA(train_daily, order=best_arima_order).fit()
print(arima_fit.summary())
```

## Cell 9 – ARIMA forecasts
**What:** two forecasts with the same order:
1. **Test-period forecast** – trained on train only, predicts the 365 test dates. Handed to the evaluator; **not** compared with actual values here.
2. **Next-30-days forecast** – refit on all data up to the last complete day, predicts 2018-08-04 to 2018-09-02.

Each table has the forecast and a 95% range and is saved to CSV next to the notebook.
**Note:** non-seasonal ARIMA cannot repeat the weekly cycle, so long forecasts look smooth/flat; that is normal.
**Expected output:** 2 plots + `arima_forecast_test_period.csv`, `arima_forecast_next_30_days.csv`.

```python
def make_forecast_table(fitted_result, steps):
    """Forecast + 95% interval as a table."""
    fc = fitted_result.get_forecast(steps=steps)
    table = fc.predicted_mean.to_frame("forecast_MW")
    ci = fc.conf_int(alpha=0.05)
    table["lower_95"] = ci.iloc[:, 0]
    table["upper_95"] = ci.iloc[:, 1]
    return table

# 1) test-period forecast (trained on train only)
arima_test_fc = make_forecast_table(arima_fit, steps=len(test_daily))
arima_test_fc.index = test_daily.index
arima_test_fc.index.name = "Date"
arima_test_fc.to_csv("arima_forecast_test_period.csv")

# 2) next 30 days (refit on all data up to last complete day)
full_series = daily.loc[:last_complete_day]
arima_full_fit = ARIMA(full_series, order=best_arima_order).fit()
steps_needed = (future_dates[-1] - last_complete_day).days      # includes the skipped incomplete day
arima_future_fc = make_forecast_table(arima_full_fit, steps=steps_needed).loc[future_dates]
arima_future_fc.index.name = "Date"
arima_future_fc.to_csv("arima_forecast_next_30_days.csv")

print("ARIMA test-period forecast:"); display(arima_test_fc.head())
print("ARIMA next-30-days forecast:"); display(arima_future_fc.head())

fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].plot(train_daily.iloc[-180:], label="Train (last 180 days)")
axes[0].plot(arima_test_fc["forecast_MW"], label="ARIMA forecast")
axes[0].fill_between(arima_test_fc.index, arima_test_fc["lower_95"], arima_test_fc["upper_95"], alpha=0.2)
axes[0].set_title("ARIMA - test period forecast (365 days)"); axes[0].legend()
axes[1].plot(full_series.iloc[-90:], label="History (last 90 days)")
axes[1].plot(arima_future_fc["forecast_MW"], label="ARIMA next 30 days")
axes[1].fill_between(arima_future_fc.index, arima_future_fc["lower_95"], arima_future_fc["upper_95"], alpha=0.2)
axes[1].set_title("ARIMA - next 30 days"); axes[1].legend()
for a in axes:
    a.grid(alpha=0.3); a.set_ylabel("MW")
plt.tight_layout(); plt.show()
```

## Cell 10 – Explain SARIMA (p, d, q)(P, D, Q, s)
**In simple words:** ARIMA plus a second "seasonal" ARIMA that looks back one whole season (e.g., the same weekday last week).

| Symbol | Meaning |
|---|---|
| **p, d, q** | as in ARIMA |
| **P** | how many previous *seasons* the values depend on |
| **D** | seasonal differencing (`today - same day last week`) |
| **Q** | how many previous *seasonal* errors are used |
| **s** | number of days in one repeating cycle |

```python
sarima_explain = pd.DataFrame({
    "Parameter": ["p", "d", "q", "P", "D", "Q", "s"],
    "Meaning": ["recent values used", "regular differencing", "recent errors used",
                "past seasons used", "seasonal differencing", "past seasonal errors used",
                "season length (days)"],
    "How I decide": ["tuned (grid)", "ADF test (Cell 5)", "tuned (grid)",
                     "tuned (grid)", "seasonality-strength test (Cell 11)", "tuned (grid)",
                     "ACF peak + EDA weekly finding (Cell 11)"]
})
display(sarima_explain)
```

## Cell 11 – Determine the seasonal period `s` (and `D`) from the EDA
**Evidence in the EDA above:** weekday (5,755 MW) vs weekend (5,222 MW) averages differ, the day-of-week chart shows a weekly cycle, and the hourly lag-168 (24x7 = one week) correlation is 0.76. In our **daily** series one week = 7 observations. I don't just assume it: the code measures which lag (2-30) has the strongest autocorrelation in the day-to-day changes.
**Yearly cycle (~365):** the EDA shows a winter/summer effect, but SARIMA with `s = 365` is impractical in statsmodels, so I use the weekly cycle and state the yearly effect as a limitation.
**`D`:** seasonal strength (share of variance explained by the repeating pattern); above 0.64 (common rule of thumb) -> `D = 1`, else `0`.
**Expected output:** top lags (7, 14, 21 should stand out), `s`, strength, `D`.

```python
changes = train_daily.diff().dropna()            # day-to-day changes so the slow level does not hide the pattern
acf_changes = acf(changes, nlags=60)

top_lags = sorted(range(2, 31), key=lambda k: acf_changes[k], reverse=True)[:5]
print("Lags (2-30) with strongest autocorrelation:", top_lags)
s = top_lags[0]
print("-> Chosen seasonal period s =", s, "days")

plt.figure(figsize=(12, 3.5))
plt.bar(range(1, 61), acf_changes[1:], width=0.6)
plt.axhline(band, color="red", ls="--"); plt.axhline(-band, color="red", ls="--")
for k in range(s, 61, s):
    plt.axvline(k, color="green", alpha=0.3)
plt.title(f"ACF of daily changes (green lines every {s} lags)")
plt.xlabel("Lag (days)"); plt.tight_layout(); plt.show()

decomp = seasonal_decompose(train_daily, model="additive", period=s)
resid = decomp.resid.dropna()
seas = decomp.seasonal.loc[resid.index]
strength = max(0, 1 - resid.var() / (resid + seas).var())
D = 1 if strength > 0.64 else 0
print(f"Seasonal strength = {strength:.2f}  ->  D = {D}")
print(f"\nFixed so far: d = {d}, D = {D}, s = {s}")
```

## Cell 12 – SARIMA hyperparameter tuning (needed: YES)
**Why:** four free parameters (`p, q, P, Q`) that the plots cannot fix.
**How (small):** `d, D, s` are fixed already. Search `p, q` in {0,1,2} and `P, Q` in {0,1} = **36 models**, lowest **AIC**, training data only.
**Speed:** the search uses only the last 3 years of the training data (1,095 days); Cell 13 then fits the chosen orders on the full training data.
**Expected output:** top-5 table and selected order. Runtime: several minutes.

```python
tune_data = train_daily.iloc[-1095:]        # last 3 years of TRAINING data only

# a constant is needed only when there is no differencing at all
trend_term = "c" if (d == 0 and D == 0) else None

sarima_results = []
start = time.time()
for p, q, P, Q in itertools.product([0, 1, 2], [0, 1, 2], [0, 1], [0, 1]):
    try:
        fitted = SARIMAX(tune_data,
                         order=(p, d, q),
                         seasonal_order=(P, D, Q, s),
                         trend=trend_term,
                         enforce_stationarity=False,
                         enforce_invertibility=False).fit(disp=False, maxiter=50)
        sarima_results.append({"order": (p, d, q), "seasonal_order": (P, D, Q, s), "AIC": fitted.aic})
    except Exception as e:
        print("Skipped", (p, d, q), (P, D, Q, s), "->", e)

sarima_table = pd.DataFrame(sarima_results).sort_values("AIC").reset_index(drop=True)
best_sarima_order = sarima_table.loc[0, "order"]
best_sarima_seasonal = sarima_table.loc[0, "seasonal_order"]

print(f"Tuning finished in {(time.time() - start) / 60:.1f} minutes")
display(sarima_table.head(5))
print("Selected SARIMA (p,d,q)(P,D,Q,s):", best_sarima_order, best_sarima_seasonal)
```

## Cell 13 – Build the final SARIMA model
**What:** fit SARIMA with the selected orders on the full **training** data.
**Expected output:** model summary.

```python
sarima_fit = SARIMAX(train_daily,
                     order=best_sarima_order,
                     seasonal_order=best_sarima_seasonal,
                     trend=trend_term,
                     enforce_stationarity=False,
                     enforce_invertibility=False).fit(disp=False, maxiter=200)
print(sarima_fit.summary())
```

## Cell 14 – SARIMA forecasts
**What:** same two forecasts as ARIMA: the 365-day test-period forecast (trained on train only, for the evaluator) and the next-30-days forecast (refit up to the last complete day). Saved as `sarima_forecast_test_period.csv` and `sarima_forecast_next_30_days.csv`. **No evaluation or comparison is done here.**
**Expected output:** two plots; the 30-day plot should show the weekly pattern (lower weekends).

```python
# 1) test-period forecast (trained on train only)
sarima_test_fc = make_forecast_table(sarima_fit, steps=len(test_daily))
sarima_test_fc.index = test_daily.index
sarima_test_fc.index.name = "Date"
sarima_test_fc.to_csv("sarima_forecast_test_period.csv")

# 2) next 30 days (refit on all data up to last complete day)
sarima_full_fit = SARIMAX(full_series,
                          order=best_sarima_order,
                          seasonal_order=best_sarima_seasonal,
                          trend=trend_term,
                          enforce_stationarity=False,
                          enforce_invertibility=False).fit(disp=False, maxiter=200)
sarima_future_fc = make_forecast_table(sarima_full_fit, steps=steps_needed).loc[future_dates]
sarima_future_fc.index.name = "Date"
sarima_future_fc.to_csv("sarima_forecast_next_30_days.csv")

print("SARIMA test-period forecast:"); display(sarima_test_fc.head())
print("SARIMA next-30-days forecast:"); display(sarima_future_fc.head())

fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].plot(train_daily.iloc[-180:], label="Train (last 180 days)")
axes[0].plot(sarima_test_fc["forecast_MW"], label="SARIMA forecast")
axes[0].fill_between(sarima_test_fc.index, sarima_test_fc["lower_95"], sarima_test_fc["upper_95"], alpha=0.2)
axes[0].set_title("SARIMA - test period forecast (365 days)"); axes[0].legend()
axes[1].plot(full_series.iloc[-90:], label="History (last 90 days)")
axes[1].plot(sarima_future_fc["forecast_MW"], label="SARIMA next 30 days")
axes[1].fill_between(sarima_future_fc.index, sarima_future_fc["lower_95"], sarima_future_fc["upper_95"], alpha=0.2)
axes[1].set_title("SARIMA - next 30 days"); axes[1].legend()
for a in axes:
    a.grid(alpha=0.3); a.set_ylabel("MW")
plt.tight_layout(); plt.show()

print("Done. Hand the *_test_period.csv files to the evaluation team member.")
```

## Why these parameters (fill in what your run printed)
* **d** – ADF on the daily training data (Cell 5).
* **ARIMA p, q** – lowest AIC in the 4x4 grid (Cell 7); ACF/PACF only gave the range. The EDA's `(3,1,3)` was checked, not copied.
* **s** – ACF of daily changes peaks at 7; matches the EDA weekday/weekend finding and the 168-hour lag. Yearly cycle not modelled (too heavy).
* **D** – seasonal-strength test (Cell 11).
* **SARIMA p, q, P, Q** – lowest AIC in the 36-model grid (Cell 12).

## Is tuning needed?
Yes for both: the EDA gave stationarity/seasonality hints but not `p, q, P, Q`, and daily ACF/PACF are cluttered by the weekly pattern. Grids are small (16 and 36 models), use training data only and AIC, so they are cheap and leak-free. `d`, `D`, `s` come from tests.

## Presentation script (~2 minutes)
"The EDA turned hourly consumption into a daily average and kept the last 365 days as test. I built ARIMA and SARIMA on that series. First I tested stationarity on the daily training data with ADF, which fixed the differencing order. ARIMA uses previous days' values (p), differencing (d) and previous errors (q). Because ACF/PACF didn't give one clear answer, I tuned p and q with a small grid using AIC on training data only. For SARIMA I added a seasonal part. The EDA showed weekday-versus-weekend differences and the ACF peaks every 7 days, so the seasonal period is 7. I tuned p, q, P, Q on a 36-model grid. Both models were trained before the test year and gave a 365-day forecast for evaluation plus a 30-day forecast beyond August 2018. Evaluation and comparison are done by a teammate."

## Likely viva questions
1. **What is ARIMA?** Past values (AR) + differencing (I) + past errors (MA).
2. **What is stationarity?** Stable mean/variance over time; ARIMA assumes it.
3. **How did you choose d?** ADF test on the daily training series.
4. **Why daily, not hourly?** 143k hourly points with several cycles would be extremely slow; the EDA chose daily averages.
5. **How did you choose s?** ACF peak at 7 plus the EDA weekly pattern; not assumed.
6. **Why not s = 365?** Too heavy for SARIMA; stated as a limitation.
7. **What is AIC?** Fit vs complexity score; lower is better.
8. **Why not tune on the test data?** It leaks the future.
9. **Why refit for the 30-day forecast?** To use the latest data with the same orders.
10. **Why is ARIMA's forecast flat?** No seasonal part, so it drifts to the average.
11. **Did you compare ARIMA and SARIMA?** No, that is another member's task.
12. **Data caveat?** The last day has one hourly reading, so the 30-day fit stops at the last complete day.
